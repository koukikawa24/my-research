```python
import numpy as np
import soundfile as sf
import scipy.io.wavfile

def ReadList(list_file):
    f = open(list_file, "r")
    lines = f.readlines()
    list_sig = []
    for x in lines:
        list_sig.append(x.rstrip())
    f.close()
    return list_sig
```

```python
data_folder = '/home/s226059/workspace/data/age_training/wav/train/'
wav_lst_tr = ReadList('/home/s226059/workspace/self_sinctdnn/label/training_scp.txt')
data_folder_test = '/home/s226059/workspace/data/age_training/wav/test/'
wav_lst_te = ReadList('/home/s226059/workspace/self_sinctdnn/label/testing_scp.txt')
```

```python
def count_hour(data_folder, wav_lst):
    a = 0
    for wav in wav_lst:
        [fs, signal] = scipy.io.wavfile.read(data_folder+wav)
        a += len(signal)
    return a/fs
```

```python
tr_hour = count_hour(data_folder, wav_lst_tr)
te_hour = count_hour(data_folder_test, wav_lst_te)
hour = (tr_hour + te_hour) / 60 / 60
print(str(hour) + '時間')
```
