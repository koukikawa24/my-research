```python
import numpy as np
import csv

def ReadList(list_file):
    f = open(list_file, "r")
    lines = f.readlines()
    list_sig = []
    for x in lines:
        list_sig.append(x.rstrip())
    f.close()
    return list_sig

def get_datas():
    path = '/home/s226059/workspace/self_sinctdnn/vox_gender.csv'
    with open(path, 'r', encoding='utf-8-sig') as f:
        reader = csv.reader(f)
        datas = [row for row in reader]

    return datas
datas = get_datas()
a1 = [a[1] for a in datas]
a3 = [a[3] for a in datas]
ID = a1[1:]
SEX = a3[1:]
textfile = '/home/s226059/workspace/self_sinctdnn/training_age.txt'
lists = ReadList(textfile)
ids_dict = {}
for i in range(len(ID)):
    ids_dict.update({ID[i]:SEX[i]})
```

```python
new_list = []
for list in lists:
    l = list.split('/')
    s = ids_dict[l[0]]
    new_list.append(list + ' ' + s)
```

```python
with open('training_age_sex.txt', 'w') as f:
    for d in new_list:
        f.write("%s\n" % d)
```

```python
data_sex = {}

for lst in new_list:
    k = lst.split()
    data_sex.update({k[0]: k[2]})
```

```python
np.save('training_sex_dict.npy', data_sex)
```
