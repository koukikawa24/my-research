```python
import numpy as np
import csv

def ReadList(list_file):
    f = open(list_file, "r")
    lines = f.readlines()
    list_sig = []
    for x in lines:
        list_sig.append(x.rstrip())
    f.close()
    return list_sig

# path = '/home/s226059/workspace/data/age_training/voxceleb2/label/train_list.txt'
# utt2spk = ReadList(path)
# id = [i.split(' ')[0] for i in utt2spk]
with open('utt_num_by_spk_base.csv') as f:
    reader = csv.reader(f)
    base_lst = [row for row in reader]

with open('utt_num_by_spk.csv') as f:
    reader = csv.reader(f)
    agev_lst = [row for row in reader]
```

```python
uttsum = 0
few_spk_id = []
many_spk_id = []
for lst in agev_lst:
    spk_id, utt_num = lst[0], int(lst[1])
    if utt_num > 20:
        uttsum += utt_num
        many_spk_id.append(spk_id)
    else:
        few_spk_id.append(spk_id) 

print(uttsum)
```
